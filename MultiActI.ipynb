```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Generar dataset sintético
np.random.seed(42)
n_muestras = 100
x = np.random.rand(n_muestras) * 10
epsilon = np.random.normal(0, 1, n_muestras)
y = 2 * x + 3 + epsilon
```

```python
#varianzas
varianza_x = np.var(x, ddof=1)
varianza_y = np.var(y, ddof=1)
```

```python
#media
media_x = np.mean(x)
media_y = np.mean(y)
```

```python
cov_xy = np.cov(x, y, ddof=1)[0, 1]
beta_1 = cov_xy / varianza_x
beta_0 = media_y - beta_1 * media_x
```

```python
print(f"Parámetros estimados:")
print(f"beta_0 (Intercepto): {beta_0:.4f}")
print(f"beta_1 (Pendiente): {beta_1:.4f}")
```

```python
#predicción
y_pred = beta_0 + beta_1 * x
```

```python
#calcular error
mse = np.mean((y - y_pred) ** 2)
```

```python
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - media_y) ** 2)
r2 = 1 - (ss_res / ss_tot)
```

```python
print(f"Métricas de evaluación:")
print(f"MSE: {mse:.4f}")
print(f"R^2: {r2:.4f}")
```

```python
#gráfico
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='blue', alpha=0.7, label='Datos sintéticos')
plt.plot(x, y_pred, color='red', linewidth=2, label=f'Línea MCO: y = {beta_1:.2f}x + {beta_0:.2f}')
plt.title('Regresión Lineal Simple desde Cero con NumPy')
plt.xlabel('Variable x')
plt.ylabel('Variable y')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
